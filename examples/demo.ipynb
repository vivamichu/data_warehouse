# dset — one line to load any ML dataset

`dset` is a **package manager for ML datasets**. Think `pip` for datasets.

This notebook demonstrates loading three different datasets — CIFAR-10, Food-101, and Oxford Pets — all with the same one-line API.

```python
# Install dset with HuggingFace support (for CIFAR-10, Food-101)
!pip install -q dset[hf]
```

## 1. CIFAR-10 — the classic 10-class benchmark

```python
from dset import load

ds = load("cifar-10")
img, label = ds[0]
print(f"CIFAR-10: {len(ds)} samples, image size={img.size}, label={label}")
img
```

## 2. Food-101 — 101 food categories, 101k images

```python
ds = load("food-101")
img, label = ds[0]
print(f"Food-101: {len(ds)} samples, image size={img.size}, label={label}")
img
```

## 3. Oxford-IIIT Pets — segmentation dataset

```python
import numpy as np
import matplotlib.pyplot as plt

ds = load("oxford-pets")
img, mask = ds[0]
print(f"Oxford-Pets: {len(ds)} samples, image size={img.size}, mask shape={mask.shape}")

fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(img)
axes[0].set_title("Image")
axes[0].axis("off")
axes[1].imshow(mask, cmap="tab10")
axes[1].set_title("Segmentation mask")
axes[1].axis("off")
plt.tight_layout()
plt.show()
```

## 4. PyTorch DataLoader integration

Any `dset` dataset is a standard `torch.utils.data.Dataset` — drop it straight into a `DataLoader`.

```python
import torch
from torch.utils.data import DataLoader
import torchvision.transforms.functional as TF

def collate_cifar(batch):
    imgs = torch.stack([TF.to_tensor(img) for img, _ in batch])
    labels = torch.tensor([label for _, label in batch])
    return imgs, labels

ds = load("cifar-10")
loader = DataLoader(ds, batch_size=32, shuffle=True, collate_fn=collate_cifar)
imgs, labels = next(iter(loader))
print(f"Batch: imgs={imgs.shape}, labels={labels.shape}")
```
